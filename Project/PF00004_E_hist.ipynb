```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from thesis_plots import *
apply_style()

import numpy as np
import matplotlib.pyplot as plt

with open("PF00004/PF00004_naturalenergies.txt") as f:
    energies = np.array([float(line) for line in f])

mean_E = np.mean(energies)
std_E = np.std(energies)

fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(energies, bins=100, linewidth=0.3)
ax.axvline(mean_E, color=SKY_BLUE, linestyle="--", linewidth=1.5, label=f"Mean = {mean_E:.1f}")
ax.axvspan(mean_E - std_E, mean_E + std_E, alpha=0.2, color=SKY_BLUE, label=f"±1 SD = {std_E:.1f}")
ax.set_xlabel("Energy")
ax.set_ylabel("Count")
ax.set_title("PF00004 Natural Energies")
ax.legend()
plt.tight_layout()
plt.show()
```
