```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from thesis_plots import *
apply_style()

import numpy as np
import matplotlib.pyplot as plt

with open("PF00004/PF00004_naturalenergies.txt") as f:
    energies = np.array([float(line) for line in f])

mean_E = np.mean(energies)
std_E = np.std(energies)

fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(energies, bins=100, linewidth=0.3)
ax.axvspan(mean_E - std_E, mean_E + std_E, color='grey', alpha=0.15, label=f"±1 Std Dev = {std_E:.1f}")
ax.axvline(mean_E, color='black', linestyle="-", linewidth=1.3, alpha=1, label=f"Mean = {mean_E:.1f}")
ax.axvline(mean_E - std_E, color='black', linestyle="--", linewidth=1.1, alpha=0.7)
ax.axvline(mean_E + std_E, color='black', linestyle="--", linewidth=1.1, alpha=0.7)
ax.set_xlabel("DCA Energy")
ax.set_ylabel("Count")
ax.set_title("PF00004 Natural Energies")
ax.legend()
plt.tight_layout()
plt.show()
```
