```python
from pymsaviz import MsaViz

# Point this to your downloaded seed alignment FASTA file
# pyMSAviz can read FASTA, Clustal, and Stockholm formats
mv = MsaViz(
    "PF00018.alignment.seed",       # path to your alignment file
    format = 'stockholm',
    color_scheme="Clustal",
    wrap_length=60,            # residues per row
    show_consensus=True,
    show_count=True,
)

# Add conservation bar on top
mv.add_markers([])  # initialises layout
mv.plotfig()
mv.savefig("msa_conservation.png", dpi=300)
mv.savefig("msa_conservation.pdf")
print("Done!")
```
